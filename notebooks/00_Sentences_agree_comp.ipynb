```python
from pathlib import Path
import pandas as pd

RAW_DIR = Path("../FinancialPhraseBank-v1.0")  # adjust if running from notebooks/
FILES = {
    "50agree": "Sentences_50Agree.txt",
    "66agree": "Sentences_66Agree.txt",
    "75agree": "Sentences_75Agree.txt",
    "allagree": "Sentences_AllAgree.txt",
}

def load_labels(filename):
    labels = []
    with open(RAW_DIR / filename, "r", encoding="latin-1") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            _, label = line.rsplit("@", 1)
            labels.append(label)
    return pd.Series(labels)

summary = {}
for name, filename in FILES.items():
    labels = load_labels(filename)
    counts = labels.value_counts()
    pcts = (labels.value_counts(normalize=True) * 100).round(1)
    summary[name] = {
        "total": len(labels),
        "negative_%": pcts.get("negative", 0),
        "neutral_%": pcts.get("neutral", 0),
        "positive_%": pcts.get("positive", 0),
    }

summary_df = pd.DataFrame(summary).T
print(summary_df)
```
